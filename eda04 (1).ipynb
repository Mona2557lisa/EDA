```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
titanic.pivot_table('survived', index='sex', columns='class')
```

```python
age = pd.cut(titanic['age'], [0, 18, 80])

titanic.pivot_table(
    values='survived',
    index=['sex', age],
    columns='class'
)
```

```python
titanic.pivot_table('survived',index='sex',columns=['class','alone'])
```

```python
titanic.pivot_table(index='sex',columns='class',aggfunc={'survived':sum,'fare':'mean'})
```

```python
titanic.pivot_table(index='sex',columns='class',aggfunc={'age':'min','fare':'min'})
```
